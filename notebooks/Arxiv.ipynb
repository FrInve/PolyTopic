```python
labels = df['terms'].tolist()

labels
```

```python
nan_count = 0

for i in range(len(labels)):
    if pd.isna(labels[i]):
        nan_count += 1

print("Number of NaN labels:", nan_count)
```

```python
import ast
```

```python
dict_labels = dict()

for i in range(len(labels)):

    if(pd.notna(labels[i])):
        label = ast.literal_eval(labels[i])

        for j in range(len(label)):
            if label[j] not in dict_labels:
                dict_labels[label[j]] = 1
            else:
                dict_labels[label[j]] += 1


print("Unique labels:", list(dict_labels.keys()))
print("Label counts:", list(dict_labels.values()))
```
