## Handle input data

```python
splits = {'train': 'DBPEDIA_train.csv', 'validation': 'DBPEDIA_val.csv', 'test': 'DBPEDIA_test.csv'}
df_train = pd.read_csv("hf://datasets/DeveloperOats/DBPedia_Classes/" + splits["train"])
df_validation = pd.read_csv("hf://datasets/DeveloperOats/DBPedia_Classes/" + splits["validation"])
df_test = pd.read_csv("hf://datasets/DeveloperOats/DBPedia_Classes/" + splits["test"])
```

```python
df_train.to_csv("data/raw/dbpedia_train.csv", index=False)
```

```python
df_all = pd.concat([df_train, df_validation, df_test], ignore_index=True)

df_all
```

```python
# Save dataframe to .csv file
df_all.to_csv('raw/dbpedia_all.csv', index=False)
```

```python
df = df_all
```

```python
# Drop rows where 'body' is missing or empty
df = df.dropna(subset=['text'])
df = df[df['text'].astype(str).str.strip() != '']
dataset = df
display(dataset.head())
print('DataFrame shape:', dataset.shape)
```

```python
df = df.drop_duplicates(subset=['text']).reset_index(drop=True)

df
```

```python
# Save dataframe to .csv file
df.to_csv('complete/dbpedia.csv', index=False)
```

```python
df_text = pd.read_parquet(f'processed/parquet/dbpedia_text.parquet')
```

```python
df_text
```

```python
# Print dataset statistics
avg_chars = df_text['text'].apply(lambda x: len(str(x))).mean()
print("Average number of characters for text passage:", avg_chars)
avg_words = df_text['text'].apply(lambda x: len(x.split(" "))).mean()
print("Average number of words for text passage:", avg_words)
print("Average word length:", avg_chars / avg_words)
```

```python
# Convert 'text' column into list containing documents
docs = df_text['text'].tolist()
```

```python
# Ensure all documents are strings
for i in range(len(docs)):
    docs[i] = str(docs[i])
```

```python
# Print length of the document list
print("Number of documents:", len(docs))
```
